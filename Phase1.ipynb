2.1 Implémentons la somme de deux vecteurs sans numpy

> Ajouter une citation



```python
def somme_vecteur(vectorun, vectordeux):
  if (len(vectorun) != len(vectordeux)):
    return 0
  somme = []
  for i in range (len(vectordeux)):
    somme.append(vectorun[i] + vectordeux[i])
  return somme
```

```python
somme = somme_vecteur([3,4],[1,2])
somme
```

2.2 Implémentons la différence de deux vecteurs sans numpy

```python
def diff_vecteur(vectorun, vectordeux):
  if (len(vectorun) != len(vectordeux)):
    return 0
  difference = []
  for i in range (len(vectordeux)):
    difference.append(vectorun[i] - vectordeux[i])
  return difference
```

```python
diff = diff_vecteur([3,4],[1,2])
diff
```

```python
def produit_scalaire_robuste(u, v):
    # TOUJOURS vérifier :
    if len(u) != len(v):
        raise ValueError("Dimensions différentes")
    if not u or not v:
        return 0
    if any(math.isnan(x) for x in u + v):
        return float('nan')
    
    return sum(a * b for a, b in zip(u, v))
```

### Visualisations

```python
import matplotlib.pyplot as plt
import numpy as np
import math

def visualiser_produit_scalaire(u, v, titre="Produit Scalaire"):
    """Visualise deux vecteurs et leur produit scalaire avec interprétation géométrique"""
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    
    # === GRAPHIQUE 1: Les vecteurs ===
    ax1.quiver(0, 0, u[0], u[1], angles='xy', scale_units='xy', scale=1, 
               color='red', width=0.005, label=f'u = {u}')
    ax1.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, 
               color='blue', width=0.005, label=f'v = {v}')
    
    # Calculs
    dot_product = sum(a * b for a, b in zip(u, v))
    norme_u = math.sqrt(sum(x**2 for x in u))
    norme_v = math.sqrt(sum(x**2 for x in v))
    
    # Angle entre les vecteurs
    if norme_u > 0 and norme_v > 0:
        cos_angle = dot_product / (norme_u * norme_v)
        # Limiter cos_angle entre -1 et 1 pour éviter les erreurs d'arrondi
        cos_angle = max(-1, min(1, cos_angle))
        angle_deg = math.degrees(math.acos(abs(cos_angle)))
        
        # Dessiner l'arc d'angle
        if angle_deg < 90:
            angles = np.linspace(0, math.radians(angle_deg), 50)
            arc_x = 0.5 * np.cos(angles)
            arc_y = 0.5 * np.sin(angles)
            ax1.plot(arc_x, arc_y, 'green', linewidth=2)
            ax1.text(0.3, 0.2, f'{angle_deg:.1f}°', color='green', fontsize=12)
    
    # Configuration graphique 1
    ax1.set_xlim(-1, max(u[0], v[0]) + 1)
    ax1.set_ylim(-1, max(u[1], v[1]) + 1)
    ax1.grid(True, alpha=0.3)
    ax1.set_aspect('equal')
    ax1.legend()
    ax1.set_title(f'Vecteurs u et v\nProduit scalaire = {dot_product}')
    ax1.axhline(y=0, color='k', linewidth=0.5)
    ax1.axvline(x=0, color='k', linewidth=0.5)
    
    # === GRAPHIQUE 2: Projection géométrique ===
    # Projection de u sur v
    if norme_v > 0:
        proj_scalar = dot_product / (norme_v ** 2)
        projection = [proj_scalar * v[0], proj_scalar * v[1]]
        
        # Dessiner la projection
        ax2.quiver(0, 0, u[0], u[1], angles='xy', scale_units='xy', scale=1, 
                   color='red', width=0.005, label=f'u = {u}')
        ax2.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, 
                   color='blue', width=0.005, label=f'v = {v}')
        ax2.quiver(0, 0, projection[0], projection[1], angles='xy', scale_units='xy', scale=1, 
                   color='purple', width=0.007, label=f'proj_v(u) = {[round(x,2) for x in projection]}')
        
        # Ligne de projection (perpendiculaire)
        ax2.plot([u[0], projection[0]], [u[1], projection[1]], 
                 'gray', linestyle='--', alpha=0.7)
        
        # Point de projection
        ax2.plot(projection[0], projection[1], 'purple', marker='o', markersize=8)
    
    ax2.set_xlim(-1, max(u[0], v[0]) + 1)
    ax2.set_ylim(-1, max(u[1], v[1]) + 1)
    ax2.grid(True, alpha=0.3)
    ax2.set_aspect('equal')
    ax2.legend()
    ax2.set_title('Interprétation géométrique:\nProjection de u sur v')
    ax2.axhline(y=0, color='k', linewidth=0.5)
    ax2.axvline(x=0, color='k', linewidth=0.5)
    
    # Informations textuelles
    interpretation = ""
    if dot_product > 0:
        interpretation = "Angle < 90° → Même direction générale"
    elif dot_product == 0:
        interpretation = "Angle = 90° → Orthogonaux (INDÉPENDANTS)"
    else:
        interpretation = "Angle > 90° → Directions opposées"
    
    fig.suptitle(f'{titre}\n{interpretation}', fontsize=14, y=1.02)
    plt.tight_layout()
    plt.show()
    
    # Affichage des résultats numériques
    print(f"📊 Résultats pour u={u} et v={v}:")
    print(f"   Produit scalaire: {dot_product}")
    print(f"   ||u|| = {norme_u:.2f}, ||v|| = {norme_v:.2f}")
    if norme_u > 0 and norme_v > 0:
        print(f"   Angle: {angle_deg:.1f}°")
        print(f"   cos(θ) = {cos_angle:.3f}")
    print(f"   Interprétation: {interpretation}")
    print("-" * 50)

# 🎯 Testons avec tes exemples
print("=== TEST 1: Vecteurs alignés ===")
visualiser_produit_scalaire([2, 3], [4, 1], "Cas 1: Vecteurs alignés")

print("\n=== TEST 2: Vecteurs orthogonaux ===")
visualiser_produit_scalaire([1, 2], [2, -1], "Cas 2: Vecteurs orthogonaux")

print("\n=== TEST 3: Vecteurs opposés ===")
visualiser_produit_scalaire([1, 1], [-2, -2], "Cas 3: Vecteurs opposés")
```

```python
def demo_applications_ml():
    """Montre des applications concrètes du produit scalaire en ML"""
    
    fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))
    
    # === 1. SYSTÈME DE RECOMMANDATION ===
    user_A = [5, 2, 0, 4, 1]  # Goûts musicaux
    user_B = [4, 3, 0, 5, 0]
    user_C = [1, 5, 4, 1, 3]
    
    genres = ['Rock', 'Pop', 'Jazz', 'Rap', 'Country']
    x = range(len(genres))
    
    ax1.bar([i-0.2 for i in x], user_A, 0.2, label='User A', alpha=0.7)
    ax1.bar([i for i in x], user_B, 0.2, label='User B', alpha=0.7)
    ax1.bar([i+0.2 for i in x], user_C, 0.2, label='User C', alpha=0.7)
    
    similarity_AB = sum(a*b for a,b in zip(user_A, user_B))
    similarity_AC = sum(a*b for a,b in zip(user_A, user_C))
    
    ax1.set_xticks(x)
    ax1.set_xticklabels(genres)
    ax1.set_title(f'Système de Recommandation\nSimilarité A-B: {similarity_AB} | A-C: {similarity_AC}')
    ax1.legend()
    
    # === 2. ANALYSE DE SENTIMENT ===
    mots = ['génial', 'nul', 'super', 'horrible', 'excellent']
    sentiment_positif = [1, -1, 1, -1, 1]
    phrase1 = [2, 0, 1, 0, 1]  # "Film génial, super et excellent"
    phrase2 = [0, 2, 0, 2, 0]  # "Film nul et horrible"
    
    ax2.bar([i-0.2 for i in range(len(mots))], sentiment_positif, 0.2, label='Polarité', alpha=0.7)
    ax2.bar([i for i in range(len(mots))], phrase1, 0.2, label='Phrase 1', alpha=0.7)
    ax2.bar([i+0.2 for i in range(len(mots))], phrase2, 0.2, label='Phrase 2', alpha=0.7)
    
    score1 = sum(a*b for a,b in zip(sentiment_positif, phrase1))
    score2 = sum(a*b for a,b in zip(sentiment_positif, phrase2))
    
    ax2.set_xticks(range(len(mots)))
    ax2.set_xticklabels(mots, rotation=45)
    ax2.set_title(f'Analyse de Sentiment\nScore 1: {score1} (positif) | Score 2: {score2} (négatif)')
    ax2.legend()
    
    # === 3. FEATURES ORTHOGONALES (ML) ===
    feature1 = [1, 2, 3, 4, 2]  # Âge normalisé
    feature2 = [2, -1, 0, -1, 1]  # Revenus normalisé (orthogonal à l'âge)
    feature3 = [1, 2, 3, 4, 2]  # Exactement corrélé à l'âge
    
    ax3.scatter(feature1, feature2, c='green', s=100, alpha=0.7, label='Features indépendantes')
    ax3.scatter(feature1, feature3, c='red', s=100, alpha=0.7, label='Features corrélées')
    
    correlation_12 = sum(a*b for a,b in zip(feature1, feature2))
    correlation_13 = sum(a*b for a,b in zip(feature1, feature3))
    
    ax3.set_xlabel('Feature 1 (âge)')
    ax3.set_ylabel('Feature 2/3')
    ax3.set_title(f'Indépendance des Features\nCorr 1-2: {correlation_12} | Corr 1-3: {correlation_13}')
    ax3.legend()
    ax3.grid(True, alpha=0.3)
    
    # === 4. WORD EMBEDDINGS ===
    # Simulation d'embeddings 2D pour visualisation
    mots_embeddings = {
        'roi': [0.8, 0.2],
        'reine': [0.7, 0.3],
        'homme': [0.6, 0.1],
        'femme': [0.5, 0.4],
        'chien': [0.1, 0.8],
        'chat': [0.2, 0.7]
    }
    
    for mot, embedding in mots_embeddings.items():
        ax4.scatter(embedding[0], embedding[1], s=200, alpha=0.7)
        ax4.annotate(mot, (embedding[0], embedding[1]), xytext=(5, 5), 
                    textcoords='offset points')
    
    # Calculer quelques similarités
    sim_roi_reine = sum(a*b for a,b in zip(mots_embeddings['roi'], mots_embeddings['reine']))
    sim_roi_chien = sum(a*b for a,b in zip(mots_embeddings['roi'], mots_embeddings['chien']))
    
    ax4.set_title(f'Word Embeddings (2D)\nSim(roi,reine): {sim_roi_reine:.2f} | Sim(roi,chien): {sim_roi_chien:.2f}')
    ax4.set_xlabel('Dimension 1')
    ax4.set_ylabel('Dimension 2')
    ax4.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

# Lancer la démo
demo_applications_ml()
```

```python
#Fonctions
def projection(u, v):
    ufoisv = sum(a * b for a, b in zip(u, v))
    norme_v = math.sqrt(sum(a**2 for a in v))
    denom_projection = norme_v**2
    parti_un = ufoisv / denom_projection
    return [a * parti_un for a in v]
```

### Exercice 2.3 - Décomposition orthogonale 

```python
# Décomposons u selon v

u = [5, 3]
v = [2, 1]

# Trouvons la composante parallèle
produit_scalaire_uv = sum(a*b for a, b in zip(u, v))
produit_scalaire_vv = sum(x**2 for x in v)
diviScalaire = (produit_scalaire_uv /produit_scalaire_vv)
uParallele = [diviScalaire * x for x in v] 

#uParallele est la partie de u qui est justifié par v, c'est à dire la partie qui va dans la direction de v
uPerpendiculaire = [round(a - b, 2) for a, b in zip(u, uParallele)]
#uPerpendiculaire est la partie qui est étrangère à v
```

```python
# Vérifie si les vecteurs e1 = [1, 0, 0] , e2 = [0, 1, 0] , e3 = [0, 0, 1] forment une base orthonormée
e1 = [1, 0, 0]
e2 = [0, 1, 0]
e3 = [0, 0, 1]

#Des vecteurs formes une bases orthonommées si la norme de chaque vecteur est égale à 1 et si leur produit scalaire est égale à 0
norme_e1 = math.sqrt(sum(x**2 for x in e1))
norme_e2 = math.sqrt(sum(x**2 for x in e2))
norme_e3 = math.sqrt(sum(x**2 for x in e3))

#Chaque vecteur a pour norme 1
def produit_three_scalaire(e1, e2, e3):
    return sum(a*b*c for a, b, c in zip(e1, e2, e3))

print(produit_three_scalaire(e1,e2,e3))

# E est une base orthonormée, car la norme de chaque vecteur qui le compose est de 1 et leur produit scalaire donne 0

def est_base_orthonormee(vecteurs):
    """
    Vérifie si une liste de vecteurs forme une base orthonormée
    """
    if not vecteurs:
        return False
    
    n = len(vecteurs)
    
    # 1. Vérifier que chaque vecteur a une norme de 1
    for v in vecteurs:
        norme = math.sqrt(sum(x**2 for x in v))
        if abs(norme - 1.0) > 1e-10:  # Tolérance pour erreurs numériques
            return False
    
    # 2. Vérifier l'orthogonalité (produit scalaire = 0 pour chaque paire)
    for i in range(n):
        for j in range(i+1, n):  # Éviter de tester deux fois la même paire
            produit_scalaire = sum(a*b for a, b in zip(vecteurs[i], vecteurs[j]))
            if abs(produit_scalaire) > 1e-10:  # Tolérance
                return False
    
    return True

# Tests
e1, e2, e3 = [1, 0, 0], [0, 1, 0], [0, 0, 1]
print(f"Base canonique orthonormée ? {est_base_orthonormee([e1, e2, e3])}")

# Test avec des vecteurs non orthonormés
v1, v2 = [1, 1, 0], [0, 1, 1]  # Pas normalisés
print(f"v1, v2 orthonormés ? {est_base_orthonormee([v1, v2])}")
```

## Processus de Gram-Schmidt (pour créer une base orthonormée) :

```python
def gram_schmidt(vecteurs):
    """
    Transforme une liste de vecteurs en base orthonormée
    """
    base_orthonormee = []
    
    for v in vecteurs:
        # Soustraire les projections sur tous les vecteurs précédents
        u = v.copy()
        for base_vector in base_orthonormee:
            # u = u - projection(u, base_vector)
            proj_coeff = sum(a*b for a,b in zip(u, base_vector))
            projection = [proj_coeff * x for x in base_vector]
            u = [a - b for a, b in zip(u, projection)]
        
        # Normaliser
        norme = math.sqrt(sum(x**2 for x in u))
        if norme > 1e-10:  # Éviter division par 0
            u_normalise = [x/norme for x in u]
            base_orthonormee.append(u_normalise)
    
    return base_orthonormee

# Test
vecteurs_initiaux = [[1, 1, 0], [1, 0, 1], [0, 1, 1]]
base_ortho = gram_schmidt(vecteurs_initiaux)
print(f"Base orthonormée créée ? {est_base_orthonormee(base_ortho)}")
```

## Exercice 3.3 - Produit vectoriel (3D uniquement)

```python
# Calcule u × v pour u = [1, 2, 0] et v = [3, 1, 4]

u = [1,2,0]
v = [3,1,4]

u_v = [(u[1] * v[2] - u[2] * v[1]), (u[2] * v[0] - u[0] * v[2]), (u[0] * v[1] - u[1] * v[0])]

# Vérifie que le résultat est perpendiculaire à u et v
#Pour que u_v soit perpendiculaire à u et à v, son produit scalaire avec chacun doit être égale à 0
    
```

```python
def produit_vectoriel(u, v):
    """
    Calcule le produit vectoriel u × v (uniquement en 3D)
    """
    if len(u) != 3 or len(v) != 3:
        raise ValueError("Le produit vectoriel n'existe qu'en 3D")
    
    return [
        u[1] * v[2] - u[2] * v[1],  # Composante x
        u[2] * v[0] - u[0] * v[2],  # Composante y  
        u[0] * v[1] - u[1] * v[0]   # Composante z
    ]

def verifier_orthogonalite(u_cross_v, u, v):
    """Vérifie que u×v est perpendiculaire à u et v"""
    perp_u = sum(a*b for a, b in zip(u_cross_v, u))
    perp_v = sum(a*b for a, b in zip(u_cross_v, v))
    
    print(f"(u × v) · u = {perp_u}")
    print(f"(u × v) · v = {perp_v}")
    
    return abs(perp_u) < 1e-10 and abs(perp_v) < 1e-10

# Test
u = [1, 2, 0]
v = [3, 1, 4]
u_cross_v = produit_vectoriel(u, v)
print(f"u × v = {u_cross_v}")
print(f"Orthogonal ? {verifier_orthogonalite(u_cross_v, u, v)}")
```

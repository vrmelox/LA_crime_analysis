```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = pd.read_csv('participants.csv')
```

```python
data
```

```python
hole = data.groupby('acteurs').agg('sum')
hole
```

```python
total_participants = hole['nombre de participants'].sum()
```

```python
hole['percentage'] = round(100 * hole['nombre de participants']/total_participants, 2)
```

```python
hole = pd.DataFrame(hole)
```

```python
hole.drop(columns='annee', inplace=True)
```

```python
hole.drop(columns='nombre de participants', inplace=True)
```

```python
hole
```

```python
# Données
labels = hole.index
valeurs = hole['percentage']

# Camembert
plt.pie(valeurs, labels=labels, autopct='%1.1f%%', startangle=90)
plt.axis('equal')  # Pour que le camembert soit un cercle parfait
plt.title('Répartition des composants')
plt.show()
```
